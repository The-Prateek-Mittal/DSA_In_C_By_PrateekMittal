```python
graph = {'u':['v','x'], 'v' : ['y'], 'w':['y','z'],'x':['v'], 'y':['x'],'z':['z']} #graph declaration

time = 0 # declaring the time as a global variable
vertex_attribute = {i : {'colour':'WHITE','parent':None, 'dTime':None,'fTime':None} for i in graph }

def DFS_Visit(graph,u):
    global time
    time = time + 1
    vertex_attribute[u]['dTime'] = time
    vertex_attribute[u]['colour'] = 'GRAY'
    
    for v in graph[u]:
        if vertex_attribute[v]['colour'] == 'WHITE':
            vertex_attribute[v]['parent'] = u
            DFS_Visit(graph,v)
            
    time = time + 1
    vertex_attribute[u]['fTime'] = time
    vertex_attribute[u]['colour'] = 'BLACK'
    

def DFS(graph):
    
    for u in graph:
        if vertex_attribute[u]['colour'] == 'WHITE':
            DFS_Visit(graph,u)
            
            
if __name__ == '__main__':
    DFS(graph)
    for vertex,attribute in vertex_attribute.items():
        print(f'{vertex}:{attribute}')
```
